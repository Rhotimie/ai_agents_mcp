### An LLM can't touch your disk - so we give it a *tool*

The model can only send back text. It has no ability to create a file. When we asked it to
"write it out to a markdown file", it just described a file it never created.

To actually get a file, *we* write the Python function that does the writing, and then
describe that function to the model in JSON. The model replies with "please call
`write_markdown_file` with these arguments", our code runs it, and we hand the result back.

That loop - model asks, our code runs, result goes back - is the core idea behind agents.

```python
# 1. The tool itself - just an ordinary Python function. This is the bit that really writes the file.

import json

def write_markdown_file(filename, content):
    with open(filename, "w", encoding="utf-8") as f:
        f.write(content)
    return f"Successfully wrote {len(content)} characters to {filename}"
```

```python
# 2. Describe the function to the model in JSON, so it knows the tool exists and how to call it

write_markdown_file_json = {
    "name": "write_markdown_file",
    "description": "Write text content to a markdown file on the user's local disk. "
                   "Call this whenever the user asks you to save, write out, or create a file.",
    "parameters": {
        "type": "object",
        "properties": {
            "filename": {
                "type": "string",
                "description": "The name of the file to write, for example 'agentic_solution.md'",
            },
            "content": {
                "type": "string",
                "description": "The full markdown text to write into the file",
            },
        },
        "required": ["filename", "content"],
        "additionalProperties": False,
    },
}

tools = [{"type": "function", "function": write_markdown_file_json}]
tools
```

```python
# 3. Now the same request as before - but this time we pass tools=tools

message = f"""
Given this business area:
{business_area}

and this pain point in the above business area:
{pain_point}

And this proposed Agentic AI solution to the above pain point:
{agentic_solution}

Format the above information as a markdown table with three columns: Business Area, Pain Point,
and Agentic AI Solution. The table should have a header row and one row of data.
Write it out to a markdown file called "agentic_solution.md" using the tool available to you.
"""

messages = [{"role": "user", "content": message}]

# Keep going until the model stops asking for tools and gives us a normal answer
while True:
    response = openai.chat.completions.create(
        model="gemini-3.5-flash", messages=messages, tools=tools
    )
    reply = response.choices[0].message

    if response.choices[0].finish_reason == "tool_calls":
        messages.append(reply)                      # remember what the model asked for
        for tool_call in reply.tool_calls:
            arguments = json.loads(tool_call.function.arguments)
            print(f"The model wants to call: {tool_call.function.name}({list(arguments)})")
            result = globals()[tool_call.function.name](**arguments)   # actually run it
            print(f"Result: {result}")
            messages.append({                        # hand the result back to the model
                "role": "tool",
                "tool_call_id": tool_call.id,
                "content": result,
            })
    else:
        break

markdown_file_solution = reply.content
print()
print(markdown_file_solution)
```

```python
# And here is the file that really exists on disk now - look for agentic_solution.md in the file explorer!
from IPython.display import Markdown, display
display(Markdown(open("agentic_solution.md").read()))
```
